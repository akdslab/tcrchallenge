```python
import DecoderTCR as dt
from DecoderTCR.reconstruct.hla import lookup_hla
import pandas as pd

model = "DecoderTCR-ESMC_300M"
allele = "HLA-B*27:05"
hla_a, hla_b = lookup_hla(allele)
# Public README peptide; no experimental label is implied.
pairs = pd.DataFrame([{
    "name": "public_example", "allele": allele,
    "HLA_a": hla_a, "HLA_b": hla_b, "Peptide": "LRVMMLAPF",
}])
scored = dt.score(pairs, model=model, device="cpu", with_tcr=False)
scored.to_csv("hla_only_scores.csv", index=False)
print(scored[["name", "allele", "Peptide", f"pll_{model}"]])
```
